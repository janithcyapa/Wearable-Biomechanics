```python
# @title Utilities
import cv2
import json
import os
import shutil
import subprocess
import threading
import numpy as np
import pandas as pd
import torch
from scipy import signal
import plotly.graph_objects as go
from ultralytics import YOLO

# YOLO (COCO) 17 keypoint names
KEYPOINT_NAMES = [
    "nose", "left_eye", "right_eye", "left_ear", "right_ear",
    "left_shoulder", "right_shoulder", "left_elbow", "right_elbow",
    "left_wrist", "right_wrist", "left_hip", "right_hip",
    "left_knee", "right_knee", "left_ankle", "right_ankle"
]

# Core biomechanical keypoints (excluding facial keypoints to reduce noise)
MOTION_KEYPOINTS = [
    "left_shoulder", "right_shoulder", "left_elbow", "right_elbow",
    "left_wrist", "right_wrist", "left_hip", "right_hip",
    "left_knee", "right_knee", "left_ankle", "right_ankle"
]

# Pre-parsed skeleton lookup array for generate_annotated_video
SKELETON_INDICES = np.array([
    (0, 1), (0, 2), (1, 3), (2, 4), (5, 6), (5, 7), (7, 9),
    (6, 8), (8, 10), (5, 11), (6, 12), (11, 12), (11, 13),
    (13, 15), (12, 14), (14, 16)
], dtype=np.int32)

# Dynamically configure device & precision so it works with or without CUDA
HAS_CUDA = torch.cuda.is_available()
INFERENCE_DEVICE = 0 if HAS_CUDA else "cpu"
USE_HALF = True if HAS_CUDA else False
DEFAULT_BATCH_SIZE = 32 if HAS_CUDA else 8

if HAS_CUDA:
    print(f"✓ CUDA active: {torch.cuda.get_device_name(0)} (batch={DEFAULT_BATCH_SIZE}, half={USE_HALF})")
else:
    print(f"ℹ Running in CPU mode (device='cpu', batch={DEFAULT_BATCH_SIZE}, half=False)")

# Initialize YOLO model once with locks for thread-safe execution
yolo_model = YOLO('yolo11x-pose.pt')
yolo_lock = threading.Lock()
store_lock = threading.Lock()


# -------------------------------------------------------------------------
# 1. Resample 60fps -> 30fps
# -------------------------------------------------------------------------
def ensure_30fps_video(video_path, target_fps=30.0):
    """
    Checks if a video is recorded above 45fps (e.g. 60fps).
    If so, resamples it down to 30fps in-place using FFmpeg ultrafast.
    Returns (video_path, was_resampled).
    """
    if not video_path or not os.path.exists(video_path):
        return None, False

    cap = cv2.VideoCapture(video_path)
    fps = cap.get(cv2.CAP_PROP_FPS) or 30.0
    cap.release()

    if fps > 45.0:
        print(f"⚠ Detected {fps:.2f} fps in {os.path.basename(video_path)}. Resampling down to {target_fps:.0f} fps...")
        temp_path = video_path.replace('.mp4', '_30fps_tmp.mp4')
        cmd = (
            f"ffmpeg -y -threads 4 -i \"{video_path}\" "
            f"-vf \"fps=fps={target_fps}\" "
            f"-c:v libx264 -preset ultrafast -crf 18 -an \"{temp_path}\""
        )
        res = subprocess.run(cmd, shell=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        if res.returncode == 0 and os.path.exists(temp_path):
            shutil.move(temp_path, video_path)
            print(f"✓ Resampled to {target_fps:.0f} fps: {video_path}")
            return video_path, True
        else:
            print(f"⚠ FFmpeg resample failed for {video_path}, keeping original.")
            if os.path.exists(temp_path):
                os.remove(temp_path)
            return video_path, False

    return video_path, False


# -------------------------------------------------------------------------
# 2. Keypoint Extraction (Largest Foreground Person Selection)
# -------------------------------------------------------------------------
def analyze_keypoints(video_path, batch_size=None, min_conf=0.25):
    """
    Thread-safe batched inference using Ultralytics generator stream.
    Selects the largest foreground person (by bounding box / keypoint span area)
    to prevent background bystanders from hijacking side-profile estimations.
    """
    if batch_size is None:
        batch_size = DEFAULT_BATCH_SIZE
    elif not HAS_CUDA and batch_size > 8:
        batch_size = 8

    all_frames_data = []

    with yolo_lock:
        results_generator = yolo_model.predict(
            source=video_path,
            batch=batch_size,
            stream=True,
            verbose=False,
            device=INFERENCE_DEVICE,
            half=USE_HALF
        )

        for frame_idx, results in enumerate(results_generator):
            frame_data = {"frame": frame_idx}

            if results.keypoints is not None and len(results.keypoints.xy) > 0:
                kpts_all = results.keypoints.xy.cpu().numpy()   # (N, 17, 2)
                confs = results.keypoints.conf.cpu().numpy()    # (N, 17)
                num_people = len(kpts_all)

                if num_people == 1:
                    best_person_idx = 0
                else:
                    # Compute bounding box area for each detected person
                    if results.boxes is not None and len(results.boxes.xywh) == num_people:
                        xywh = results.boxes.xywh.cpu().numpy()
                        areas = xywh[:, 2] * xywh[:, 3]
                        box_confs = results.boxes.conf.cpu().numpy()
                    else:
                        # Fallback: compute area from visible keypoint span
                        areas = np.zeros(num_people, dtype=np.float32)
                        box_confs = np.max(confs, axis=1)
                        for p_idx in range(num_people):
                            valid_pts = kpts_all[p_idx][confs[p_idx] >= min_conf]
                            valid_pts = valid_pts[(valid_pts[:, 0] > 0) & (valid_pts[:, 1] > 0)]
                            if len(valid_pts) >= 2:
                                w = np.max(valid_pts[:, 0]) - np.min(valid_pts[:, 0])
                                h = np.max(valid_pts[:, 1]) - np.min(valid_pts[:, 1])
                                areas[p_idx] = max(w, 10.0) * max(h, 10.0)

                    # Gate out low-confidence ghost detections, then pick largest area
                    top6_kp_conf = np.mean(np.sort(confs, axis=1)[:, -6:], axis=1)
                    valid_person_mask = (box_confs >= 0.30) & (top6_kp_conf >= min_conf)

                    if np.any(valid_person_mask):
                        masked_areas = np.where(valid_person_mask, areas, -1.0)
                        best_person_idx = int(np.argmax(masked_areas))
                    else:
                        best_person_idx = int(np.argmax(areas))

                kpts = kpts_all[best_person_idx]
                scores = confs[best_person_idx]

                if len(kpts) == 17:
                    for i, name in enumerate(KEYPOINT_NAMES):
                        x = float(kpts[i][0])
                        y = float(kpts[i][1])
                        s = float(scores[i])
                        if s < min_conf or (x == 0.0 and y == 0.0):
                            frame_data[f"{name}_x"] = np.nan
                            frame_data[f"{name}_y"] = np.nan
                            frame_data[f"{name}_score"] = s
                        else:
                            frame_data[f"{name}_x"] = x
                            frame_data[f"{name}_y"] = y
                            frame_data[f"{name}_score"] = s
                else:
                    for name in KEYPOINT_NAMES:
                        frame_data[f"{name}_x"] = np.nan
                        frame_data[f"{name}_y"] = np.nan
                        frame_data[f"{name}_score"] = 0.0
            else:
                for name in KEYPOINT_NAMES:
                    frame_data[f"{name}_x"] = np.nan
                    frame_data[f"{name}_y"] = np.nan
                    frame_data[f"{name}_score"] = 0.0

            all_frames_data.append(frame_data)

    df = pd.DataFrame(all_frames_data)
    output_json_path = video_path.replace('.mp4', '_keypoints.json')
    df.to_json(output_json_path, orient='records')
    print(f"Keypoints saved to {output_json_path}")
    return df


# -------------------------------------------------------------------------
# 3. Calculate Temporal Sync Lag (Front vs Side)
# -------------------------------------------------------------------------
def calculate_sync_lag(front_df, side_df, plot=False):
    """
    Calculates the frame lag between the front and side videos.
    Lag = t_front - t_side:
      - Positive lag (> 0): Front video started earlier; trim first `best_lag` frames of front video.
      - Negative lag (< 0): Side video started earlier; trim first `|best_lag|` frames of side video.
    """
    def get_movement_signal(df):
        x_cols = [f"{name}_x" for name in MOTION_KEYPOINTS]
        y_cols = [f"{name}_y" for name in MOTION_KEYPOINTS]
        s_cols = [f"{name}_score" for name in MOTION_KEYPOINTS]

        kx = df[x_cols].copy()
        ky = df[y_cols].copy()
        ks = df[s_cols].fillna(0.0).to_numpy()

        low_conf_mask = ks < 0.3
        kx[low_conf_mask] = np.nan
        ky[low_conf_mask] = np.nan

        kx = kx.interpolate(method='linear', limit_direction='both').ffill().bfill().fillna(0.0).to_numpy()
        ky = ky.interpolate(method='linear', limit_direction='both').ffill().bfill().fillna(0.0).to_numpy()

        dx = np.diff(kx, axis=0)
        dy = np.diff(ky, axis=0)

        pair_conf = np.minimum(ks[:-1], ks[1:])
        joint_speed = np.sqrt(0.25 * dx**2 + dy**2) * (pair_conf >= 0.3)

        raw_sig = np.sum(joint_speed, axis=1)
        if len(raw_sig) >= 5:
            raw_sig = signal.medfilt(raw_sig, kernel_size=5)
        return raw_sig

    sig_front = get_movement_signal(front_df)
    sig_side  = get_movement_signal(side_df)

    sig_front_norm = (sig_front - np.mean(sig_front)) / (np.std(sig_front) + 1e-8)
    sig_side_norm  = (sig_side - np.mean(sig_side)) / (np.std(sig_side) + 1e-8)

    correlation = signal.correlate(sig_front_norm, sig_side_norm, mode='full')
    lags = signal.correlation_lags(len(sig_front_norm), len(sig_side_norm), mode='full')

    best_lag = int(lags[np.argmax(correlation)])

    if plot:
        fig1 = go.Figure()
        fig1.add_trace(go.Scatter(y=sig_front_norm, mode='lines', name='Front Movement', opacity=0.8))
        fig1.add_trace(go.Scatter(y=sig_side_norm, mode='lines', name='Side Movement', opacity=0.8))
        fig1.update_layout(title="Original Un-Synced Movement Signals", template="plotly_dark", height=400)
        fig1.show()

        fig2 = go.Figure()
        if best_lag > 0:
            x1 = np.arange(len(sig_front_norm))
            x2 = np.arange(len(sig_side_norm)) + best_lag
        else:
            x1 = np.arange(len(sig_front_norm)) - best_lag
            x2 = np.arange(len(sig_side_norm))

        fig2.add_trace(go.Scatter(x=x1, y=sig_front_norm, mode='lines', name='Front Movement (Synced)', opacity=0.8))
        fig2.add_trace(go.Scatter(x=x2, y=sig_side_norm, mode='lines', name='Side Movement (Synced)', opacity=0.8))
        fig2.update_layout(title=f"Aligned Movement Signals (Side Lag: {best_lag} frames)", template="plotly_dark", height=400)
        fig2.show()

    return best_lag


# -------------------------------------------------------------------------
# 4. Generate Skeleton-Annotated Video
# -------------------------------------------------------------------------
def generate_annotated_video(video_path, keypoints_df):
    """
    High-speed vectorized skeleton drawing with fast encoding and NaN checks.
    """
    CONFIDENCE_THRESHOLD = 0.4
    cap = cv2.VideoCapture(video_path)
    width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps    = cap.get(cv2.CAP_PROP_FPS) or 30.0

    out_path = video_path.replace('.mp4', '_annotated.mp4')
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(out_path, fourcc, fps, (width, height))

    total_frames = min(int(cap.get(cv2.CAP_PROP_FRAME_COUNT)), len(keypoints_df))
    x_cols = [f"{name}_x" for name in KEYPOINT_NAMES]
    y_cols = [f"{name}_y" for name in KEYPOINT_NAMES]
    s_cols = [f"{name}_score" for name in KEYPOINT_NAMES]

    all_kpts_x = keypoints_df[x_cols].to_numpy()
    all_kpts_y = keypoints_df[y_cols].to_numpy()
    all_scores = keypoints_df[s_cols].to_numpy()

    frame_idx = 0
    while cap.isOpened() and frame_idx < total_frames:
        ret, frame = cap.read()
        if not ret:
            break

        scores = all_scores[frame_idx]
        kx = all_kpts_x[frame_idx]
        ky = all_kpts_y[frame_idx]
        valid_mask = (scores >= CONFIDENCE_THRESHOLD) & (~np.isnan(kx)) & (~np.isnan(ky))

        for p1, p2 in SKELETON_INDICES:
            if valid_mask[p1] and valid_mask[p2]:
                pt1 = (int(kx[p1]), int(ky[p1]))
                pt2 = (int(kx[p2]), int(ky[p2]))
                cv2.line(frame, pt1, pt2, (0, 255, 0), 2, lineType=cv2.LINE_4)

        for i in np.where(valid_mask)[0]:
            cv2.circle(frame, (int(kx[i]), int(ky[i])), 4, (0, 0, 255), -1)

        out.write(frame)
        frame_idx += 1

    cap.release()
    out.release()
    print(f"Annotated video saved to {out_path}")
    return out_path


# -------------------------------------------------------------------------
# 5. Generate Synced Side-by-Side Video (Supports Missing Camera Half-Black)
# -------------------------------------------------------------------------
def generate_synced_video(front_video_path, side_video_path, side_lag_frames=None):
    """
    Frame-accurate side-by-side video generation using multi-threaded FFmpeg.
    - If both videos exist: trims leading frames using `side_lag_frames` and hstacks (Left=Front, Right=Side).
    - If only front_video exists: places Front on Left and pads Right half with black.
    - If only side_video exists: places Side on Right and pads Left half with black.
    """
    has_front = bool(front_video_path and os.path.exists(front_video_path))
    has_side  = bool(side_video_path and os.path.exists(side_video_path))

    if not has_front and not has_side:
        print("Error: No valid video paths provided to generate_synced_video.")
        return None

    base_dir = os.path.dirname(front_video_path if has_front else side_video_path)
    out_path = os.path.join(base_dir, 'synced_side_by_side.mp4')
    if os.path.exists(out_path):
        os.remove(out_path)

    ref_path = front_video_path if has_front else side_video_path
    cap = cv2.VideoCapture(ref_path)
    fps = cap.get(cv2.CAP_PROP_FPS) or 30.0
    cap.release()

    if has_front and has_side:
        lag = int(side_lag_frames) if side_lag_frames is not None else 0
        if lag > 0:
            trim_f = f"trim=start_frame={lag},setpts=PTS-STARTPTS,"
            trim_s = "setpts=PTS-STARTPTS,"
        elif lag < 0:
            trim_f = "setpts=PTS-STARTPTS,"
            trim_s = f"trim=start_frame={abs(lag)},setpts=PTS-STARTPTS,"
        else:
            trim_f = "setpts=PTS-STARTPTS,"
            trim_s = "setpts=PTS-STARTPTS,"

        filter_complex = (
            f"[0:v]{trim_f}fps={fps:.4f},scale=-2:720[v0];"
            f"[1:v]{trim_s}fps={fps:.4f},scale=-2:720[v1];"
            f"[v0][v1]hstack=inputs=2:shortest=1[v]"
        )
        cmd = (
            f"ffmpeg -y -threads 4 "
            f"-i \"{front_video_path}\" "
            f"-i \"{side_video_path}\" "
            f"-filter_complex \"{filter_complex}\" "
            f"-map \"[v]\" -c:v libx264 -preset ultrafast -crf 23 -an \"{out_path}\""
        )
    elif has_front and not has_side:
        filter_complex = f"[0:v]setpts=PTS-STARTPTS,fps={fps:.4f},scale=-2:720,pad=iw*2:ih:0:0:black[v]"
        cmd = (
            f"ffmpeg -y -threads 4 "
            f"-i \"{front_video_path}\" "
            f"-filter_complex \"{filter_complex}\" "
            f"-map \"[v]\" -c:v libx264 -preset ultrafast -crf 23 -an \"{out_path}\""
        )
    else:
        filter_complex = f"[0:v]setpts=PTS-STARTPTS,fps={fps:.4f},scale=-2:720,pad=iw*2:ih:iw:0:black[v]"
        cmd = (
            f"ffmpeg -y -threads 4 "
            f"-i \"{side_video_path}\" "
            f"-filter_complex \"{filter_complex}\" "
            f"-map \"[v]\" -c:v libx264 -preset ultrafast -crf 23 -an \"{out_path}\""
        )

    result = subprocess.run(cmd, shell=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    if result.returncode == 0 and os.path.exists(out_path):
        print(f"Synced video saved to {out_path}")
        return out_path
    else:
        print("FFmpeg synchronization fallback: check video input paths.")
        return None
```

```python
# @title Visual Estimation Pipeline
import os
from concurrent.futures import ThreadPoolExecutor

def visual_estimation_pipeline(entry_id, store, generate_annotated=True, generate_synced=True, plot=True):
    """
    Pipeline for visual estimation of a specific experiment.
    Gracefully handles missing front or side videos, resamples 60fps videos to 30fps,
    and generates single-view half-black synced videos when one camera is missing.
    """
    entry_type = "experiments"

    exp_folder = f"/content/{entry_id}"
    os.makedirs(exp_folder, exist_ok=True)
    print(f"Setting up directory: {exp_folder}")

    def safe_download(label):
        try:
            path = store.download_related_file(
                entry_type=entry_type,
                entry_id=entry_id,
                label=label,
                dest_folder=exp_folder
            )
            if path and os.path.exists(path):
                return path
        except Exception as e:
            print(f"ℹ {label} not available for {entry_id} ({e})")
        return None

    print("Checking/Downloading front_video...")
    front_video_path = safe_download("front_video")

    print("Checking/Downloading side_video...")
    side_video_path = safe_download("side_video")

    has_front = front_video_path is not None
    has_side  = side_video_path is not None

    if not has_front and not has_side:
        print(f"Error: Neither front_video nor side_video found for {entry_id}.")
        return None, None, None, None, None, None

    # 1. Normalize FPS to 30Hz before keypoint extraction & syncing
    front_resampled = False
    side_resampled  = False

    if has_front:
        front_video_path, front_resampled = ensure_30fps_video(front_video_path, target_fps=30.0)
        if front_resampled:
            with store_lock:
                try:
                    store.add_related_file(
                        entry_type=entry_type,
                        entry_id=entry_id,
                        label="front_video",
                        local_path=front_video_path
                    )
                    print(f"✓ Updated 30fps front_video in OpenDatasetStore for {entry_id}")
                except Exception as e:
                    print(f"⚠ Could not overwrite resampled front_video in store: {e}")

    if has_side:
        side_video_path, side_resampled = ensure_30fps_video(side_video_path, target_fps=30.0)
        if side_resampled:
            with store_lock:
                try:
                    store.add_related_file(
                        entry_type=entry_type,
                        entry_id=entry_id,
                        label="side_video",
                        local_path=side_video_path
                    )
                    print(f"✓ Updated 30fps side_video in OpenDatasetStore for {entry_id}")
                except Exception as e:
                    print(f"⚠ Could not overwrite resampled side_video in store: {e}")

    # 2. Analyze Keypoints for available video(s)
    print("\n--- Analyzing Keypoints ---")
    front_df = analyze_keypoints(front_video_path, batch_size=32) if has_front else None
    side_df  = analyze_keypoints(side_video_path, batch_size=32) if has_side else None

    # 3. Calculate Sync Lag only when both videos are present
    side_lag = None
    if has_front and has_side:
        print("\n--- Calculating Sync Lag ---")
        side_lag = calculate_sync_lag(front_df, side_df, plot=plot)
        print(f"Calculated side lag: {side_lag} frames")
    else:
        missing_cam = "side_video" if has_front else "front_video"
        print(f"\n--- Skipping Sync Lag Calculation ({missing_cam} is missing) ---")

    # 4. Generate Annotated Video(s)
    front_annotated = None
    side_annotated  = None
    synced_video    = None

    if generate_annotated:
        print("\n--- Generating Annotated Video(s) ---")
        with ThreadPoolExecutor(max_workers=2) as render_pool:
            f_front = render_pool.submit(generate_annotated_video, front_video_path, front_df) if has_front else None
            f_side  = render_pool.submit(generate_annotated_video, side_video_path, side_df) if has_side else None
            if f_front:
                front_annotated = f_front.result()
            if f_side:
                side_annotated = f_side.result()

    # 5. Generate Synced Video (or single-camera half-black video)
    if generate_synced:
        print("\n--- Generating Synchronized Side-by-Side Video (FFmpeg Ultrafast) ---")
        synced_video = generate_synced_video(front_video_path, side_video_path, side_lag)

    print("\nPipeline complete!")
    return front_df, side_df, side_lag, front_annotated, side_annotated, synced_video
```

```python
# @title Save Results
def save_visual_estimation_results(entry_id, store, front_df, side_df, side_lag,
                                   front_annotated_path=None, side_annotated_path=None,
                                   synced_video_path=None):
    """
    Thread-safe saving of results from the visual estimation pipeline back to the OpenDatasetStore.
    """
    entry_type = "experiments"
    with store_lock:
        print(f"--- Saving results for {entry_id} to OpenDatasetStore ---")

        # 1. Save Metadata (Lag Frames)
        if side_lag is not None:
            store.edit_entry(
                entry_type=entry_type,
                entry_id=entry_id,
                **{"video_sync_side_lag_frames": int(side_lag)}
            )
            print(f"✓ Saved metadata: video_sync_side_lag_frames = {int(side_lag)}")

        # 2. Save Keypoint DataFrames
        if front_df is not None and not front_df.empty:
            store.add_processed_data(
                entry_type=entry_type,
                entry_id=entry_id,
                tag="front_video_keypoints",
                df=front_df
            )
            print("✓ Saved front_video_keypoints dataframe")

        if side_df is not None and not side_df.empty:
            store.add_processed_data(
                entry_type=entry_type,
                entry_id=entry_id,
                tag="side_video_keypoints",
                df=side_df
            )
            print("✓ Saved side_video_keypoints dataframe")

        # 3. Save Generated Videos
        def save_video(file_path, label):
            if file_path and os.path.exists(file_path):
                try:
                    store.add_related_file(
                        entry_type=entry_type,
                        entry_id=entry_id,
                        label=label,
                        local_path=file_path
                    )
                    print(f"✓ Saved video file for label: {label}")
                except AttributeError:
                    print(f"⚠ Could not automatically save {label}. Store API might require a different method.")
                except Exception as e:
                    print(f"⚠ Error saving {label}: {e}")

        save_video(synced_video_path, "synced_video")
        save_video(front_annotated_path, "front_annotated_video")
        save_video(side_annotated_path, "side_annotated_video")

        print("--- Save Complete! ---")
```

```python
# @title Parallel Run: Visual Estimation & Auto-Save
import time
import shutil
from concurrent.futures import ThreadPoolExecutor, as_completed

experiments_to_run = [
    "experiments_0023",
    "experiments_0024",
    "experiments_0025",
    "experiments_0026",
    "experiments_0027",
    "experiments_0028",
    "experiments_0029",
    "experiments_0030",
    "experiments_0031",
    "experiments_0032",
    "experiments_0033",
    "experiments_0034",
    "experiments_0035",
    "experiments_0036",
    "experiments_0037",
]

def run_single_experiment(entry_id):
    """Worker to process and save a single experiment."""
    print(f"\n[START] Pipeline started for: {entry_id}")
    t0 = time.time()

    # 1. Run the visual estimation pipeline (plot=False in parallel batch mode)
    results = visual_estimation_pipeline(
        entry_id=entry_id,
        store=store,
        generate_annotated=True,
        generate_synced=True,
        plot=True
    )

    if not results or (results[0] is None and results[1] is None):
        print(f"\n[FAIL] Processing failed for: {entry_id} (no valid videos found)")
        return entry_id, False

    front_df, side_df, side_lag, front_annotated_path, side_annotated_path, synced_video_path = results

    # 2. Save back using thread-safe save function
    save_visual_estimation_results(
        entry_id=entry_id,
        store=store,
        front_df=front_df,
        side_df=side_df,
        side_lag=side_lag,
        front_annotated_path=front_annotated_path,
        side_annotated_path=side_annotated_path,
        synced_video_path=synced_video_path
    )

    # Optional: Clean up local /content/{entry_id} scratch directory to prevent Colab disk full errors
    exp_folder = f"/content/{entry_id}"
    if os.path.exists(exp_folder):
        shutil.rmtree(exp_folder, ignore_errors=True)

    dt = time.time() - t0
    print(f"\n[DONE] Finished & saved {entry_id} in {dt:.1f}s")
    return entry_id, True

MAX_WORKERS = 3

print(f"Launching batch execution with {MAX_WORKERS} parallel workers...")
start_all = time.time()

with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
    futures = {executor.submit(run_single_experiment, exp_id): exp_id for exp_id in experiments_to_run}

    for future in as_completed(futures):
        exp_id = futures[future]
        try:
            _, success = future.result()
        except Exception as e:
            print(f"[ERROR] Exception during processing of {exp_id}: {e}")

print(f"\n{'='*50}\nAll experiments completed in {time.time() - start_all:.1f}s\n{'='*50}")
```

# Raw Sync Pipeline

```python
# @title
import os
import cv2
import numpy as np
import pandas as pd
from scipy import signal
import plotly.graph_objects as go

def calculate_front_to_raw_sync_lag(
    front_df,
    raw_df,
    front_video_path=None,
    video_fps=None,
    imu_fs=100.0,
    max_lag_sec=25.0,
    plot=True
):
    """
    Robustly synchronizes Front Video keypoints (auto-detecting 30Hz vs 60Hz)
    with 100Hz Raw IMU Accelerometer data.

    Lag Definition:
      Delta_t = t_front - t_raw
      - Positive delta_t (> 0): Front video started recording earlier than raw IMU data.
      - Negative delta_t (< 0): Raw IMU data started recording earlier than front video.
    """
    # -------------------------------------------------------------------------
    # 1. Extract 1-D movement energy signal from Front Video Keypoints
    # -------------------------------------------------------------------------
    MOTION_KEYPOINTS = [
        "left_shoulder", "right_shoulder", "left_elbow", "right_elbow",
        "left_wrist", "right_wrist", "left_hip", "right_hip",
        "left_knee", "right_knee", "left_ankle", "right_ankle"
    ]
    x_cols = [f"{name}_x" for name in MOTION_KEYPOINTS]
    y_cols = [f"{name}_y" for name in MOTION_KEYPOINTS]
    s_cols = [f"{name}_score" for name in MOTION_KEYPOINTS]

    kx = front_df[x_cols].copy()
    ky = front_df[y_cols].copy()
    ks = front_df[s_cols].fillna(0.0).to_numpy()

    low_conf_mask = ks < 0.3
    kx[low_conf_mask] = np.nan
    ky[low_conf_mask] = np.nan

    kx = kx.interpolate(method='linear', limit_direction='both').ffill().bfill().fillna(0.0).to_numpy()
    ky = ky.interpolate(method='linear', limit_direction='both').ffill().bfill().fillna(0.0).to_numpy()

    dx = np.diff(kx, axis=0, prepend=kx[:1])
    dy = np.diff(ky, axis=0, prepend=ky[:1])
    pair_conf = np.minimum(ks, np.roll(ks, 1, axis=0))
    pair_conf[0] = ks[0]

    joint_speed = np.sqrt(0.25 * dx**2 + dy**2) * (pair_conf >= 0.3)
    sig_front_raw = np.sum(joint_speed, axis=1)
    if len(sig_front_raw) >= 5:
        sig_front_raw = signal.medfilt(sig_front_raw, kernel_size=5)
    # Clip extreme camera-bump / edge dropout spikes (like the end of newplot(1))
    sig_front_raw = np.clip(sig_front_raw, 0, np.percentile(sig_front_raw, 99.5))

    # -------------------------------------------------------------------------
    # 2. Extract 1-D dynamic movement signal from Raw IMU Accelerometer (100Hz)
    # -------------------------------------------------------------------------
    accel_cols = [c for c in raw_df.columns if c.endswith(('_AX', '_AY', '_AZ'))]
    accel_data = raw_df[accel_cols].interpolate(method='linear', limit_direction='both').fillna(0.0).to_numpy()

    n_sensors = accel_data.shape[1] // 3
    accel_3d = accel_data[:, :n_sensors * 3].reshape(-1, n_sensors, 3)

    accel_mag = np.linalg.norm(accel_3d, axis=2)
    gravity_baseline = np.median(accel_mag, axis=0, keepdims=True)
    dynamic_accel = np.abs(accel_mag - gravity_baseline)

    sig_imu_100 = np.sum(dynamic_accel, axis=1)
    if len(sig_imu_100) >= 5:
        sig_imu_100 = signal.medfilt(sig_imu_100, kernel_size=5)
    sig_imu_100 = np.clip(sig_imu_100, 0, np.percentile(sig_imu_100, 99.5))

    t_imu_100 = np.arange(len(sig_imu_100)) / float(imu_fs)
    imu_duration_s = t_imu_100[-1]

    # -------------------------------------------------------------------------
    # 3. Auto-Detect Video FPS (30Hz vs 60Hz) via Video File or Signal Ratio
    # -------------------------------------------------------------------------
    detected_fps = None
    if video_fps is not None:
        detected_fps = float(video_fps)
    elif front_video_path and os.path.exists(front_video_path):
        cap = cv2.VideoCapture(front_video_path)
        raw_fps = cap.get(cv2.CAP_PROP_FPS)
        cap.release()
        if raw_fps and raw_fps > 0:
            detected_fps = 60.0 if raw_fps > 45.0 else 30.0

    # Helper to evaluate alignment at a candidate FPS (30.0 or 60.0)
    def evaluate_at_fps(cand_fps):
        # Downsample 100Hz IMU onto candidate video FPS grid
        t_grid = np.arange(0, imu_duration_s, 1.0 / cand_fps)
        sig_imu_grid = np.interp(t_grid, t_imu_100, sig_imu_100)

        # Smooth both with a 2.0 Hz low-pass filter for robust repetition-level correlation
        b, a = signal.butter(3, 2.0 / (0.5 * cand_fps), btype='low')
        sf_smooth = signal.filtfilt(b, a, sig_front_raw) if len(sig_front_raw) > 15 else sig_front_raw
        si_smooth = signal.filtfilt(b, a, sig_imu_grid) if len(sig_imu_grid) > 15 else sig_imu_grid

        # Z-score normalize both smoothed envelopes (for correlation) and raw signals (for plotting)
        sf_corr = (sf_smooth - np.mean(sf_smooth)) / (np.std(sf_smooth) + 1e-8)
        si_corr = (si_smooth - np.mean(si_smooth)) / (np.std(si_smooth) + 1e-8)

        sf_plot = (sig_front_raw - np.mean(sig_front_raw)) / (np.std(sig_front_raw) + 1e-8)
        si_plot = (sig_imu_grid - np.mean(sig_imu_grid)) / (np.std(sig_imu_grid) + 1e-8)

        corr = signal.correlate(sf_corr, si_corr, mode='full')
        lags = signal.correlation_lags(len(sf_corr), len(si_corr), mode='full')

        # Restrict search to physically plausible lag window: |lag| <= max_lag_sec
        max_lag_frames = int(round(max_lag_sec * cand_fps))
        valid_mask = np.abs(lags) <= max_lag_frames
        if not np.any(valid_mask):
            valid_mask = np.ones_like(lags, dtype=bool)

        valid_lags = lags[valid_mask]
        valid_corr = corr[valid_mask]

        # Apply a mild 10% center-preference taper to break ties between adjacent repetitions
        taper = 1.0 - 0.10 * (np.abs(valid_lags) / (max_lag_frames + 1e-8))
        weighted_corr = valid_corr * taper

        best_idx = int(np.argmax(weighted_corr))
        best_lag_f = int(valid_lags[best_idx])

        # Compute true normalized correlation coefficient r in [-1, 1] on the overlapping segment
        norm_denom = np.sqrt(np.sum(sf_corr**2) * np.sum(si_corr**2)) + 1e-8
        peak_r = float(valid_corr[best_idx] / norm_denom)

        return best_lag_f, peak_r, sf_plot, si_plot

    if detected_fps is not None:
        best_lag_frames, corr_r, sig_front_norm, sig_raw_norm = evaluate_at_fps(detected_fps)
        final_fps = detected_fps
    else:
        # Test both 30Hz and 60Hz and combine correlation score with duration-ratio prior
        inferred_fps_ratio = len(front_df) / max(imu_duration_s, 1.0)
        res_30 = evaluate_at_fps(30.0)
        res_60 = evaluate_at_fps(60.0)

        # If frame-to-duration ratio is > 44 frames/sec, strongly favor 60Hz
        score_30 = res_30[1] * (1.25 if inferred_fps_ratio <= 44.0 else 0.75)
        score_60 = res_60[1] * (1.25 if inferred_fps_ratio > 44.0 else 0.75)

        if score_60 > score_30:
            final_fps = 60.0
            best_lag_frames, corr_r, sig_front_norm, sig_raw_norm = res_60
        else:
            final_fps = 30.0
            best_lag_frames, corr_r, sig_front_norm, sig_raw_norm = res_30

    best_lag_sec = best_lag_frames / final_fps
    best_lag_imu_samples = int(round(best_lag_sec * imu_fs))

    # -------------------------------------------------------------------------
    # 4. Interactive Diagnostic Plot
    # -------------------------------------------------------------------------
    if plot:
        fig = go.Figure()
        fig.add_trace(go.Scatter(
            x=np.arange(len(sig_front_norm)),
            y=sig_front_norm,
            mode='lines',
            name=f'Front Video Movement ({int(final_fps)}Hz)',
            opacity=0.85
        ))

        x_raw_shifted = np.arange(len(sig_raw_norm)) + best_lag_frames
        fig.add_trace(go.Scatter(
            x=x_raw_shifted,
            y=sig_raw_norm,
            mode='lines',
            name=f'Raw IMU Accel Movement (Aligned {int(final_fps)}Hz)',
            opacity=0.85
        ))
        fig.update_layout(
            title=(
                f"Aligned Movement Signals — Accelerometer | Detected Video: {int(final_fps)}Hz | "
                f"Lag: {best_lag_frames:+d} frames ({best_lag_sec:+.3f}s) | r = {corr_r:.3f}"
            ),
            xaxis_title=f"Front Video Frame Index ({int(final_fps)}Hz)",
            yaxis_title="Normalized Movement Energy (Z-score)",
            template="plotly_dark",
            height=400
        )
        fig.show()

    return {
        "video_fps": final_fps,
        "delta_t_frames": best_lag_frames,
        "delta_t_imu_samples": best_lag_imu_samples,
        "delta_t_sec": round(best_lag_sec, 4),
        "corr_peak": round(corr_r, 4)
    }
```
