```python
import opensim as osim
import os

print("Setting up Static Optimization Pipeline...")

# 1. Define File Names (Replace these with the files you uploaded to Colab)
model_file = "your_musculoskeletal_model.osim"
kinematics_file = "your_imu_joint_angles.mot"
results_directory = "SO_Results"

# Create a folder to store the output data
if not os.path.exists(results_directory):
    os.makedirs(results_directory)

# 2. Load the Musculoskeletal Model
model = osim.Model(model_file)
model.initSystem()

# 3. Setup the Analyze Tool
# The AnalyzeTool is the engine that runs various analyses, including Static Optimization
analyze_tool = osim.AnalyzeTool()
analyze_tool.setModel(model)
analyze_tool.setName("IMU_Muscle_Tension_Analysis")

# Load your IMU derived joint angles
analyze_tool.setCoordinatesFileName(kinematics_file)

# 4. Configure Static Optimization
so = osim.StaticOptimization()
so.setName("StaticOptimization")

# By default, Static Optimization attempts to minimize the sum of squared muscle activations.
so.setUseMusclePhysiology(True)

# Set the start and end times to match your IMU data's timeline
# (Update these values to match the timestamps in your .mot file)
initial_time = 0.0
final_time = 2.0

so.setStartTime(initial_time)
so.setEndTime(final_time)
analyze_tool.setInitialTime(initial_time)
analyze_tool.setFinalTime(final_time)

# Add the optimization routine to our tool
analyze_tool.addAnalysis(so)

# 5. External Loads (Crucial for weightlifting analysis)
# If your subject is holding a weight, the model needs to know.
# You would uncomment this and provide an XML file defining the barbell mass and where it is being held.
# analyze_tool.setExternalLoadsFileName("barbell_external_forces.xml")

# 6. Run the Simulation
analyze_tool.setResultsDir(results_directory)

print(f"Running Static Optimization from t={initial_time} to t={final_time}...")
analyze_tool.run()

print(f"Success! Muscle tension data has been saved to the '{results_directory}' folder.")
```
